## <b>Thesis</b> <i> - การวิเคราะห์ข้อมูลสถิติ Generative AI</i>

```python
from bert_score import score
```

#### การวิเคราะห์ข้อมูลด้วย BERTScore

```python
cands = ["คุณมีอาการปวดหัวมากหรือไม่"]
refs = ["คุณมีอาการปวดหัวรุนแรงหรือไม่?"]
P, R, F1 = score(
  cands, refs, 
  model_type = "airesearch/wangchanberta-base-att-spm-uncased",
  num_layers=12,
  lang="th", 
  verbose=True)
print("P", P.tolist(), "R", R.tolist(), "F1", F1.tolist())
```
